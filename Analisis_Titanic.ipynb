```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Cargar el dataset del Titanic
df = pd.read_csv('train.csv')
df.head()
```

```python
# Gráfica de sobrevivientes según el sexo
sns.countplot(data=df, x='Survived', hue='Sex')
plt.title('Supervivencia en el Titanic por Género')
plt.xlabel('0 = No sobrevivió, 1 = Sí sobrevivió')
plt.ylabel('Cantidad de pasajeros')
plt.show()
```

```python
# Gráfica de supervivencia por clase de pasajero
sns.countplot(data=df, x='Pclass', hue='Survived')
plt.title('Supervivencia en el Titanic según la Clase del Pasajero')
plt.xlabel('Clase (1 = 1era, 2 = 2da, 3 = 3era)')
plt.ylabel('Cantidad de pasajeros')
plt.show()
```
